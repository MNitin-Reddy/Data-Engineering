* Look at the data types here the Salary is int64 lbelow is the list of space taken by memory for the int types# Look at the data types here the Salary is int64 lbelow is the list of space taken by memory for the int types
- int8 - 1 byte -128 to 127
- int16 - 2 bytes -32768 to 32767
- int32 - 4 bytes -2.1 billion to 2.1 billion

- float32 - 4 bytes -  7 decimal

Salaray is int64 - taking more space if the data fits in int32 convert to save the data space 


#### Data pipelines have finite memory not all can fit into the RAM, processing may become slow or fail


```python
df["Salary"] = df["Salary"].astype("Int32")
```

```python
# investige and then decide 

# Boundaries 
print("min value Salary:", df["Salary"].min() , "\n", "max value Salary:", df["Salary"].max())

print(f"Bonus % boundaries min: { df['Bonus %'].min() }, max: {df['Bonus %'].max()}")

print(df['Salary'].nunique())
print(df['Bonus %'].nunique())


print(df.memory_usage(deep=True).sort_values(ascending=False))
```

```python
for col in df.select_dtypes(include = 'object').columns:
    print(f"\ncol: {col}")
    print("unique values:", df[col].nunique())
    print(df[col].value_counts(dropna = False ).head(10))
```

```python
#checking the memory usage 

before = df.memory_usage(deep = True).sum()
print(f"Before: {before}")

team = df['Team'].astype("category")

print("original memory:", df['Team'].memory_usage(deep = True))
print("Category", team.memory_usage(deep = True))
```

#### loc - loc[row number, column] - use : to represent all - looks specifically at the dataframes index

```python
print(df.loc[2, "Team"])

#print(df.loc[:, ["Team", "First Name"]].head(10) )

# print(df.loc[20:30 , :]) # Both Inclusive

print(df.loc[[1,5], ["Team", "First Name"]]) # index starts from 0
```

#### iloc iloc[row number, column number] - unlike loc which selects based on Labels this one selects based on position
Excludes the ending number


```python
print(df.iloc[0:5, 2:4])
# Returns 5 rows and just 2 columns as the ending values are not included
```

* WHen loading a csv pd already assigns a number index to the dataset and loc works on that index - if the data is sorted the index values remains jumbled in the sort 

### Boolean Filtering 

Boolean mask - the boolean series obtained when run the code below

```python
df['Salary'] > 50000
```

```python
mask = df['Salary'] > 0

filtered_df = df[mask]

filtered_df.tail(4)
```

```python
print( df[ df["Salary"]> 100000 ][["First Name", "Gender"]].head(5)  )

print(df[~df["Team"].isin(["Finance", "Product"]) ].head(2))

print(df[(df["Team"]== "Finance") & (df["Salary"] > 100000)].head(1))
```

#### Group by and Named Aggregations

```python
gp = df.groupby("Team")
# This creates a group by object

print(gp.size())
print('---'*10)

print(gp.ngroups)
```

```python
# Multiple Aggregations 
df.groupby(["Team", "Senior Management"])["Salary"].agg(['mean', 'min', 'max', 'count']).head(2)

## Size counts nulls too but count only takes non nulls
```

```python
# Named Aggregations 

df.groupby("Team", dropna = False, as_index = False).agg(
        avg_salary=("Salary", "mean"),
        max_salary=("Salary", "max"),
        count_=("Salary", "count")
)
```

```python
df["team_avg_salary"] = (
    df.groupby("Team")["Salary"].transform("sum")
)
```
